## 1. Applying Custom Functions with `apply()` / `map()`
- `apply()` lets you apply a custom function across rows or columns.
- `map()` is for applying functions element-wise to a Series.
- Both are handy for transformations, calculations, and cleaning data.

```python
import pandas as pd 
import numpy as np 
data = {
    "Name": ["Alice", "Bob", "Charlie", "David", "Emma"],
    "Math": [88, 75, 92, 60, 85],
    "Science": [90, 80, 78, 70, 95]
}

df = pd.DataFrame(data)
print(df)
df["Total"] = df[["Math","Science"]].apply(lambda row:row.sum(),axis=1)
print(df)
```

```python
col_mean = df[["Math","Science"]].apply(lambda row:row.mean(),axis=1)

col_mean
```

```python
#convert the score  to grade

def to_grade(s):
  if s >= 90:
    return "A"
  elif s >=80:
    return "B"
  elif s >= 70:
    return "C"
  else:
    return "D"

df["Science_Grade"] = df["Science"].apply(to_grade)
df["Math_Grade"] = df["Math"].apply(to_grade)
df
```

```python
df["Name_lower"] = df["Name"].map(str.lower)
```

```python
df["Math"] = df["Math"].map(lambda x : x+5)
```

```python
df["Gender"] = ["F","M","M","F","F"]
df
```

```python
df["Gender"] = df["Gender"].map({
    "F" :0,
    "M" :1

})
df
```

```python
import pandas as pd

data = {
    "Name": ["Alice", "Bob", "Charlie", "David", "Eva", "Frank", "Grace"],
    "Age": [23, 35, 29, 40, 22, 31, 28],
    "City": ["Delhi", "Mumbai", "Delhi", "Pune", "Mumbai", "Delhi", "Pune"],
    "Salary": [50000, 80000, 60000, 90000, 45000, 70000, 65000],
    "Department": ["IT", "HR", "IT", "Finance", "HR", "IT",
                   "Finance"]
}

df = pd.DataFrame(data)
print(df)
```

```python
# Low (<50k), Medium (50k–70k), High (>70k)

def level(s):
  if s <=50000:
    return "Low"
  elif s <=70000:
    return "Medium"
  elif s <=70000:
    return "High"

df["Level_salary"]  = df["Salary"].apply(level)
df
```

```python
# 👉 Delhi → DL, Mumbai → MB, Pune → PN

df["City"] = df["City"].replace({
    "Delhi":"Dl",
    "Mumbai":"MB",
    "Pune":"PN"
})
df
```

```python
# IT → Tech, HR → Human Resource, Finance → Fin
df["Department"] = df["Department"].replace({
    "IT":"Tech",
    "HR":"Human resource",
    "Finance":"FIn"
})
df
```

```python
# 👉 Young (<25), Adult (25–35), Senior (>35)

def level(s):
  if s<25:
    return "Young"
  elif s<=35:
    return "Adult"
  elif s>=35:
    return "Senior"

df["Level_Age"] = df["Age"].apply(level)
df
```

```python
# To update salaries based on city, using apply with a lambda function or a custom function is more appropriate.
# The KeyError is because df["City","Salary"] is an incorrect way to select multiple columns.
# It should be df[["City","Salary"]] to select a DataFrame with those two columns.
# However, .map() is for Series. For row-wise operations involving multiple columns, .apply(axis=1) is used.

# Let's assume the goal is to add 500 to the salary if the city is 'Dl'.
def update_salary_based_on_city(row):
  if row["City"] == "Dl":
    return row["Salary"] + 500
  else:
    return row["Salary"] # Keep original salary for other cities

df["Update_salary"] = df.apply(update_salary_based_on_city, axis=1)
df
```
