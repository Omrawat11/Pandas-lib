## 1. Converting Columns to Datetime (`pd.to_datetime`)
- Use `pd.to_datetime()` to convert strings or numbers into datetime objects.
- Once converted, Pandas recognizes them as dates, enabling powerful operations.

```python
import pandas as pd

# 1. Create the dataset
data = {
    "OrderID": [1001, 1002, 1003, 1004, 1005],
    "OrderDate": ["2024-01-05", "2024/01/12", "15-01-2024", "2024-01-20", "20240125"],
    "Customer": ["Alice", "Bob", "Charlie", "David", "Emma"],
    "Amount": [250, 400, 150, 300, 500]
}

df = pd.DataFrame(data)
```

```python
df
```

```python
df.info()
```

```python
df['OrderDate'] = pd.to_datetime(df['OrderDate'],errors="coerce",format="mixed")
```

## 2. Setting DateTime Index & Resampling
- Set a datetime column as the index with `set_index()`.
- Resample time series using `resample('M').mean()` or other rules ('D', 'W', 'Y').
- Useful for aggregating data by time periods.

```python
  # 1- Convert OrderDate to datetime 
df["OrderDate"] = pd.to_datetime(df["OrderDate"],errors="coerce",format="mixed")

# 2- Set OrderDate as index 
df_indx = df.set_index("OrderDate")
print(df_indx)

# 3- resample by month to compute total amount 
# Fixed: Changed from resample("ME")("Amount") to resample("ME")["Amount"]
monthly_total = df_indx.resample("ME")["Amount"].sum()
print(monthly_total) 

# 4- Resample by week to compute averge amount
weekly_total = df_indx.resample("W")["Amount"].sum()
print(weekly_total)

w_avg = df_indx.resample("W")["Amount"].mean()
print(w_avg)
```

```python
import pandas as pd

# 1️⃣ Create a dataset of website visits (string dates + visit counts)
data = {
    "VisitDate": [
        "2024-01-03", "2024/01/08", "15-01-2024",
        "2024-02-02", "2024-02-15",
        "2024/03/01", "2024-03-12", "01-04-2024"
    ],
    "Visitors": [120, 150, 90, 200, 180, 220, 210, 300]
}

df = pd.DataFrame(data)

# 2️⃣ Convert VisitDate to datetime
# Write your code here
df["VisitDate"] = pd.to_datetime(df["VisitDate"],errors = "coerce",format="mixed")
print(df)

# 3️⃣ Set VisitDate as the index
# Write your code here
visiter_index=df.set_index("VisitDate")
print(visiter_index)
# 4️⃣ Resample to get total visitors per month
# Write your code here

print(visiter_index.resample("M")["Visitors"].sum())

# 5️⃣ Resample to get average visitors per week
# Write your code here
print(visiter_index.resample("W")["Visitors"].mean())
```
